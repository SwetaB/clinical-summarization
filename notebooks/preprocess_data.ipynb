```python
import pandas as pd
```

```python
df_raw = pd.read_csv("../data/raw/PMC-Patients.csv")
df_raw.dropna(inplace=True)
df_raw['row_len'] = df_raw['patient'].apply(lambda x: len(str(x)))
df_subset = df_raw[df_raw['row_len']< 1500]
print(df_subset.shape)
df_subset.to_csv("../data/raw/PMC-Patients-Subset-ls1500.csv", index=False)
```
